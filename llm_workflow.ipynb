```python
from langgraph.graph import StateGraph,START,END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
class definestate(TypedDict):
    prompt:str
    outline:str
    response:str
    result:float
```

```python
def outline_generator(state:definestate)->definestate:
    prompt=state['prompt']
    llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
    outline_answer=f"Generate a detailed outline for {prompt}"
    state['outline']=llm.invoke(outline_answer).content

    return state
    
```

```python
def response_generator(state:definestate)->definestate:
    prompt=state['prompt']
    outline=state['outline']
    response_answer=f"Generate a detailed blog on {prompt} and outline for it is{outline}"
    state['response']=llm.invoke(response_answer).content

    return state
    
```

```python
def result_generator(state:definestate)->definestate:
    prompt=state['prompt']
    outline=state['outline']
    response=state['response']
    result_answer=f"Rate this {response} on the basis of its {prompt} and its {outline}"
    state['result']=llm.invoke(result_answer).content

    return state
```

```python
graph=StateGraph(definestate)
graph.add_node("outline_generator",outline_generator)
graph.add_node("response_generator",response_generator)
graph.add_node("result_generator",result_generator)

#add edges
graph.add_edge(START,"outline_generator")
graph.add_edge("outline_generator","response_generator")
graph.add_edge("response_generator","result_generator")
graph.add_edge("result_generator",END)

workflow=graph.compile()
```

```python
initial_state={"prompt":"Which is better framework to build agentic ai LangGrapg,Langchain,n8n,make.com and why?"}
final_state=workflow.invoke(initial_state)


print(final_state['outline'])

```
